# Project - 9 Data Visualization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class DataAnalyzer:

    def __init__(self):
        self.data = pd.DataFrame()
```

1. Load Dataset............

```python
def load_data(self):

    print("\n== Load Dataset ==")

    file_name = input("\nEnter the path of the dataset (CSV file): ")
    self.data = pd.read_csv(file_name)

    print("\nDataset loaded successfully!")

DataAnalyzer.load_data = load_data
```

2. Explore Data...............

```python
def explore_data(self):

    while True:

        print("\n== Explore Data ==")
        print("1. Display the first 5 rows")
        print("2. Display the last 5 rows")
        print("3. Display column names")
        print("4. Display data types")
        print("5. Display basic info")
        print("6. Back to Main Menu")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            print("\nFirst 5 Rows:")
            print(self.data.head())

        elif choice == "2":
            print("\nLast 5 Rows:")
            print(self.data.tail())

        elif choice == "3":
            print("\nColumn Names:")
            print(self.data.columns.tolist())

        elif choice == "4":
            print("\nData Types:")
            print(self.data.dtypes)

        elif choice == "5":
            print("\nBasic Information:")
            self.data.info()

        elif choice == "6":
            break

        else:
            print("Invalid choice! Please try again.")
            
DataAnalyzer.explore_data = explore_data
```

3. Perform Dataframe Operations.............. 

```python
def df_operations(self):

    while True:

        print("\n== DataFrane Operations ==")
        print("1. Display dataset shape")
        print("2. Select specific columns")
        print("3. Filter Data")
        print("4. Sort Data")
        print("5. Add calculated column")
        print("6. Back to Main Menu")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            print("\nDataset Shape:")
            print(self.data.shape)

        elif choice == "2":
            print("\nSelected Columns:")
            print(self.data[["Product","Units Sold","Total Sale"]].head())

        elif choice == "3":
            filtered_data = self.data[self.data["Total Sale"] >= 50000]

            print("\nFiltered Data:")
            print(filtered_data)

        elif choice == "4":
            sorted_data = self.data.sort_values(by="Total Sale", ascending=True)

            print("\nSorted Data:")
            print(sorted_data.head())

        elif choice == "5":
            self.data["Tax"] = self.data["Total Sale"] * 0.10

            print("\nCalculated Column 'Tax' Added:")
            print(self.data[["Total Sale", "Tax"]].head())

        elif choice == "6":
            break

        else:
            print("Invalid choice! Please try again.")

DataAnalyzer.df_operations = df_operations
```

4. Handle Missing Data..............

```python
def handle_data(self):

    while True:

        print("\n== Handle Missing Data ==")
        print("1. Display rows with missing values")
        print("2. Fill missing values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")
        print("5. Back to Main Menu")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            print("\nRows with missing values:")
            print(self.data[self.data.isnull().any(axis=1)])

        elif choice == "2":
            self.data["Units Sold"] = self.data["Units Sold"].fillna(self.data["Units Sold"].mean())
            self.data["Unit Price"] = self.data["Unit Price"].fillna(self.data["Unit Price"].mean())
            self.data["Total Sale"] = self.data["Total Sale"].fillna(self.data["Total Sale"].mean())

            print("\nMissing numerical value filled with mean")

        elif choice == "3":
            self.data.dropna(inplace=True)
            print("\nRows with missing values dropped")

        elif choice == "4":
            value = input("Enter value to replace missing value: ")
            self.data = self.data.fillna(value)

            print("\nMissing valuesa replaced successfully")

        elif choice == "5":
            break

        else:
            print("Invaid choice! Please try again")

DataAnalyzer.handle_data = handle_data
```

5. Generate Descriptive Statistics........... 

```python
def descriptive_statistics(self):

    while True:

        print("\n== Generate Descriptive Statistics ==")
        print("1. Display descriptive statistics")
        print("2. Calculate mean")
        print("3. Calculate median")
        print("4. Calculate standard deviation")
        print("5. Calculate variance")
        print("6. Back to Main Menu")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            print("\nDescriptive Statistics:")
            print(self.data.describe())

        elif choice == "2":
            print("\nMean of Total Sale:")
            print(self.data["Total Sale"].mean())

        elif choice == "3":
            print("\nMedian of Total Sale:")
            print(self.data["Total Sale"].median())

        elif choice == "4":
            print("\nStandard Deviation of Total Sale:")
            print(self.data["Total Sale"].std())

        elif choice == "5":
            print("\nVariance of Total Sale:")
            print(self.data["Total Sale"].var())

        elif choice =="6":
            break

        else:
            print("Inalid choice! Please try again")

DataAnalyzer.descriptive_statistics = descriptive_statistics
```

6. Data Visualization...............

```python
def data_visualization(self):

    while True:

        print("\n== Data Visualization ==")
        print("1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Stack plot")
        print("7. Box Plot")
        print("8. Heatmap")
        print("9. Back to Main Menu")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            self.data.groupby("Product")["Total Sale"].sum().plot(kind="bar")

            plt.title("Total Sales by Product")
            plt.xlabel("Product")
            plt.ylabel("Total Sale")
            plt.show()

        elif choice == "2":
            self.data.groupby("Sale Date")["Total Sale"].sum().plot()

            plt.title("Sale Trend")
            plt.xlabel("Sale Date")
            plt.ylabel("Total Sale")
            plt.show()

        elif choice == "3":
            plt.scatter(self.data["Units Sold"],
                        self.data["Total Sale"])
            plt.title("Units Sold vs Total Sale")
            plt.xlabel("Units Sold")
            plt.ylabel("Total Sale")
            plt.show()

        elif choice == "4":
            self.data.groupby("Sales Channel")["Total Sale"].sum().plot(kind="pie", autopct="%1.1f%%")

            plt.title("Sales by Sales Channel")
            plt.ylabel("")
            plt.show()

        elif choice == "5":
            self.data["Total Sale"].dropna().plot(kind = "hist",bins=10)

            plt.title("Distribution of Total Sale")
            plt.xlabel("Total Sale")
            plt.show()

        elif choice == "6":
            data = self.data.groupby(["Sale Date","Sales Channel"])["Total Sale"].sum().unstack()

            data.plot(kind="area",stacked=True)

            plt.title("Stack Plot - Sales by Channel")
            plt.xlabel("Sale Date")
            plt.ylabel("Total Sale")
            plt.show()

        elif choice == "7":
            plt.figure(figsize=(12,6))
            self.data.boxplot(column = "Total Sale", by="State")

            plt.title("Total Sale by State")
            plt.suptitle("")
            plt.xlabel("State")
            plt.ylabel("Total Sale")
            plt.xticks(rotation=45)
            plt.show()

        elif choice == "8":
            data = self.data[
                ["Units Sold","Unit Price","Total Sale"]
            ].corr()

            sns.heatmap(data,annot=True)
            plt.title("Correlation Heatmap")
            plt.show()

        elif choice == "9":
            break

        else:
            print("Invalid choice! Please try again")

DataAnalyzer.data_visualization = data_visualization
```

7. Save Visualization...........

```python
def save_visualization(self):

    print("\n== Save Visualization ==")

    file_name = input("Enter the name to save the plot (e.g., scatter_plot.png): ")

    plt.savefig("graph" + file_name)

    print("Visualization saved as", file_name, "successfully!")

DataAnalyzer.save_visualization = save_visualization
```

# Main Menu.....

```python
def menu(self):

    while True:

        print("\n========== Data Analysis & Visualization Program ==========")
        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")
        print("=============================================================")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            self.load_data()

        elif choice == "2":
            self.explore_data()

        elif choice == "3":
            self.df_operations()

        elif choice == "4":
            self.handle_data()

        elif choice == "5":
            self.descriptive_statistics()

        elif choice == "6":
            self.data_visualization()

        elif choice == "7":
            self.save_visualization()
            
        elif choice == "8":

            print("\nExiting the program. Goodbye!")
            break

        else:
            print("Invalid choice! Please try again.")

DataAnalyzer.menu = menu
```

```python
analyzer = DataAnalyzer()
analyzer.menu()
```
